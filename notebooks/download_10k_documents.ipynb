# Download Documents

```python
import os
import sys

from edgar import Company, set_identity

sys.path.insert(0, "..")
import config
from dotenv import load_dotenv
load_dotenv("../.env")

EDGAR_IDENTITY = os.getenv("EDGAR_IDENTITY", "student@edu.rtu.lv")
set_identity(EDGAR_IDENTITY)

DOCUMENTS_DIR = config.DOCUMENTS_DIR
DOCUMENTS_DIR.mkdir(parents=True, exist_ok=True)
print(f"Saving to: {DOCUMENTS_DIR}")
print(f"EDGAR identity: {EDGAR_IDENTITY}")
```

```python
COMPANIES = {
    "apple": "AAPL",
    "nvidia": "NVDA",
    "amd": "AMD",
    "intel": "INTC",
    "microsoft": "MSFT",
}

for company_key, ticker in COMPANIES.items():
    print(f"\n{company_key.upper()} ({ticker})")
    company = Company(ticker)
    filing = company.get_filings(form="10-K").latest()
    print(f"Filing date: {filing.filing_date}")

    out_path = DOCUMENTS_DIR / f"{company_key}_10k.txt"
    try:
        text = filing.text()
        out_path.write_text(text, encoding="utf-8")
        print(f"Saved: {out_path.name}")
    except Exception as e:
        print(f"ERROR: {e}")

print("\nDone.")
```

```python
files = sorted(DOCUMENTS_DIR.glob("*.txt"))
print(f"Files in {DOCUMENTS_DIR}:")
for f in files:
    size_kb = f.stat().st_size // 1024
    print(f"{f.name}: {size_kb} KB")
```
